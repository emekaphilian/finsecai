```python

from typing import Union
from pathlib import Path
import pandas as pd

```

```python
# --------------------------
# Ensure a column is datetime
# --------------------------
def ensure_ts(df: pd.DataFrame, col: str) -> pd.DataFrame:
    df = df.copy()
    if col not in df.columns:
        df[col] = pd.Timestamp.now()
    df[col] = pd.to_datetime(df[col], errors='coerce')
    df.dropna(subset=[col], inplace=True)
    return df

# --------------------------
# Parse human-readable timedelta (e.g., '5m', '2h', '1d') to pd.Timedelta
# --------------------------
def parse_time_delta(value: Union[str, pd.Timedelta]) -> pd.Timedelta:
    if isinstance(value, pd.Timedelta):
        return value
    try:
        unit = value[-1]
        num = int(value[:-1])
        if unit == 'm':
            return pd.Timedelta(minutes=num)
        elif unit == 'h':
            return pd.Timedelta(hours=num)
        elif unit == 'd':
            return pd.Timedelta(days=num)
        else:
            return pd.Timedelta(minutes=num)
    except Exception:
        return pd.Timedelta(minutes=5)

# --------------------------
# Save DataFrame to CSV and return path
# --------------------------
def save_df(df: pd.DataFrame, path: Union[str, Path], index: bool = False) -> Path:
    p = Path(path)
    p.parent.mkdir(parents=True, exist_ok=True)
    df.to_csv(p, index=index, encoding='utf-8')
    return p

if __name__ == "__main__":
    print(f"✓ pipeline_utils.py executed successfully")

```
